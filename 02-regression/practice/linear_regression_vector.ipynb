```python
import numpy as np
```

```python
xi = [453, 11, 86]
w0 = 7.17
w = [0.01,0.04,0.002]
```

```python
def dot(xi, w):
    m = len(xi)
    res = 0.0
    for i in range(m):
        res += xi[i] * w[i]
    return res
```

```python
def linear_regression(xi):
    return w0 + dot(xi, w)
```

```python
linear_regression(xi)
```

```python
w_new= [w0] + w
xi = [1, 453, 11, 86]
```

```python
def linear_regression(xi):
    return dot(xi,w_new)
```

```python
linear_regression(xi)
```
